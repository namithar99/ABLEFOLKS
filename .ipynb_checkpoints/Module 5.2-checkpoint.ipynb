```python
#Summary statistics
import pandas as pd
import numpy as np
from scipy import stats

# Sample dataset
data = pd.DataFrame({
    'Marks': [35, 50, 65, 70, 85, 90, 95]
})

# Central tendency
mean_val = data['Marks'].mean()
median_val = data['Marks'].median()
mode_val = data['Marks'].mode()[0]

# Spread
variance_val = data['Marks'].var()
std_dev_val = data['Marks'].std()
range_val = data['Marks'].max() - data['Marks'].min()

# Other stats
min_val = data['Marks'].min()
max_val = data['Marks'].max()
q1 = data['Marks'].quantile(0.25)
q3 = data['Marks'].quantile(0.75)
percentile_90 = np.percentile(data['Marks'], 90)

print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
print("Variance:", variance_val)
print("Standard Deviation:", std_dev_val)
print("Range:", range_val)
print("Min:", min_val)
print("Max:", max_val)
print("Q1 (25%):", q1)
print("Q3 (75%):", q3)
print("90th Percentile:", percentile_90)
```

```python
#Checking normality/skewness.
import numpy as np
from scipy.stats import skew, kurtosis, shapiro

np.random.seed(42)
delivery_times = np.random.exponential(scale=30, size=200)   # Skewed data

print("Skewness:", skew(delivery_times))
print("Kurtosis:", kurtosis(delivery_times))

stat,p = shapiro(delivery_times)
print("\nShapiro-Wilk Test Statistic:", stat)
print("p-value:", p)

if p > 0.05:
    print("Data looks normally distributed")
else:
    print("Data is not normally distributed")
```

```python
import numpy as np
from scipy.stats import skew, kurtosis, kstest

np.random.seed(1)
scores = np.random.normal(loc=65, scale=12, size=180)

print("Skewness:", skew(scores))
print("Kurtosis:", kurtosis(scores))

stat, p = kstest(scores, 'norm', args=(scores.mean(), scores.std()))
print("\nK-S Test Statistic:", stat)
print("p-value:", p)

if p > 0.05:
    print("Data looks normally distributed")
else:
    print("Data is not normally distributed")
```

```python
#Index 0 → 15% significance ,Index 1 → 10% significance, Index 2 → 5% significance, Index 3 → 2.5% significance, Index 4 → 1% significance
from scipy.stats import anderson
np.random.seed(10)
spending = np.random.lognormal(mean=3.5, sigma=0.8, size=250)

print("Skewness:", skew(spending))
print("Kurtosis:", kurtosis(spending))

result = anderson(spending)

print("\nAnderson-Darling Statistic:", result.statistic)
print("Critical Values:", result.critical_values)
print("Significance Levels:", result.significance_level)

if result.statistic < result.critical_values[2]:  # 5% level
    print("Data looks normally distributed (5% significance)")
else:
    print("Data is not normally distributed (5% significance)")
```

```python
#Trend & Time Series Analysis
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Date": pd.date_range(start="2024-01-01", periods=12, freq="M"),
    "Sales": [200, 220, 215, 230, 240, 250, 245, 260, 270, 280, 290, 310]
}
df = pd.DataFrame(data)
df.set_index("Date", inplace=True)

# Line plot
plt.figure(figsize=(8,5))
plt.plot(df.index, df["Sales"], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

```python
#Chi sqaure
import pandas as pd
from sklearn.feature_selection import chi2
from sklearn.preprocessing import LabelEncoder

data = {
    "Gender": ["Male","Female","Male","Female","Male","Female","Male","Female"],
    "AgeGroup": ["Young","Young","Middle","Middle","Senior","Senior","Young","Middle"],
    "Income": ["Low","Medium","High","Medium","Low","High","Medium","Low"],
    "Purchase": ["Yes","No","Yes","Yes","No","Yes","No","Yes"]
}
df = pd.DataFrame(data)
print("Dataset:\n", df)

# Encode categorical variables into numbers
le = LabelEncoder()
for col in df.columns:
    df[col] = le.fit_transform(df[col])
print("\nEncoded Dataset:\n", df)

# Separate features and target
X = df.drop("Purchase", axis=1)  # Features
y = df["Purchase"]              # Target    Y=a+bX

#  Apply Chi-Square test
chi_scores, p_values = chi2(X, y)
for feature, score, p in zip(X.columns, chi_scores, p_values):
    print(f"Feature: {feature} | Chi2 Score: {score:.2f} | p-value: {p:.4f}")
```

```python
#ANOVA
import pandas as pd
from sklearn.feature_selection import f_classif
from sklearn.preprocessing import LabelEncoder
data = {
    "Math_Score": [88, 92, 80, 60, 45, 78, 95, 65, 72, 50],
    "Science_Score": [85, 90, 78, 55, 40, 75, 92, 60, 70, 48],
    "English_Score": [75, 70, 82, 65, 55, 72, 80, 60, 68, 50],
    "Result": ["Pass", "Pass", "Pass", "Fail", "Fail", "Pass", "Pass", "Fail", "Pass", "Fail"]
}
df = pd.DataFrame(data)

# Split into features and target
X = df.drop("Result", axis=1)  # Numeric features
y = df["Result"]               # Categorical target
# Encode target (Pass/Fail → 1/0)
le = LabelEncoder()
y_encoded = le.fit_transform(y)

f_scores, p_values = f_classif(X, y_encoded)
anova_results = pd.DataFrame({"Feature": X.columns, "F-Score": f_scores, "p-Value": p_values})
print(anova_results)
```

```python
#Mutual Information
import pandas as pd
from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import LabelEncoder
data = {
    "Age": [25, 30, 22, 35, 28],
    "Gender": ["M", "F", "F", "M", "F"],
    "Income": [50000, 60000, 35000, 80000, 45000],
    "Purchased": ["Yes", "No", "Yes", "No", "Yes"]
}
df = pd.DataFrame(data)

#  Encode categorical features and target
le_gender = LabelEncoder()
df['Gender'] = le_gender.fit_transform(df['Gender'])  # M=1, F=0
le_target = LabelEncoder()
y = le_target.fit_transform(df['Purchased'])  # Yes=1, No=0
X = df.drop("Purchased", axis=1)

# Calculate Mutual Information
mi = mutual_info_classif(X, y, discrete_features=[False, True, False])  # Gender is discrete
mi_series = pd.Series(mi, index=X.columns)
mi_series = mi_series.sort_values(ascending=False)
print("Mutual Information scores:")
print(mi_series)
```

```python
#Forward selection
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Sample dataset
data = {
    "Study_Hours": [2, 4, 6, 8, 10, 12, 14, 16],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "IQ": [90, 100, 110, 105, 115, 120, 125, 130],
    "Marks": [50, 55, 60, 65, 70, 75, 80, 85]
}
df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Marks", axis=1)   # Features #inputs for prediction.
y = df["Marks"]                # Target (what we predict)

remaining = list(X.columns)    # features not yet chosen
selected = []                  # chosen features
best_r2 = 0

# Forward Selection process
while remaining:
    scores = {}               #store the performance R² score
    for feature in remaining:
        # Try adding this feature
        trial = selected + [feature]
        model = LinearRegression().fit(X[trial], y) #train the model w.r.t y
        r2 = r2_score(y, model.predict(X[trial ])) #compares the real Marks (y) vs predicted Marks
        scores[feature] = r2
    
    # Pick the best feature
    best_feature = max(scores, key=scores.get) #select the feature with highest R² score
    best_score = scores[best_feature]
    
    if best_score > best_r2:  # add only if improves score
        selected.append(best_feature)
        remaining.remove(best_feature)
        best_r2 = best_score
        print(f"Added: {best_feature}, R² = {best_r2:.3f}")
    else:
        break

print("\nFinal selected features:", selected)
print("Final R² Score:", best_r2)

# Train final model with selected features
final_model = LinearRegression().fit(X[selected], y)

# Predict Marks
predicted_marks = final_model.predict(X[selected])
df["Predicted_Marks"] = predicted_marks

print("\nPredicted Marks:\n", df)
```

```python
# Backward Elimination
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
data = {
    "Size_sqft": [1200, 1500, 1800, 2000, 2200],
    "Bedrooms": [2, 3, 3, 4, 4],
    "Age": [10, 5, 15, 10, 20],
    "Price_USD": [50000, 80000, 75000, 100000, 95000]
}
df = pd.DataFrame(data)
X = df.drop("Price_USD", axis=1)
y = df["Price_USD"]

# Backward Elimination
selected = list(X.columns)  # start with all features
best_r2 = r2_score(y, LinearRegression().fit(X[selected], y).predict(X[selected]))

while True:
    scores = {}
    for feature in selected:
        trial = selected.copy()
        trial.remove(feature)
        model = LinearRegression().fit(X[trial], y)
        r2 = r2_score(y, model.predict(X[trial]))
        scores[feature] = r2

    # Find feature whose removal increases or least decreases R²
    worst_feature = min(scores, key=scores.get)
    worst_score = scores[worst_feature]

    if worst_score >= best_r2:  # if removing improves or no change
        selected.remove(worst_feature)
        best_r2 = worst_score
        print(f"Removed: {worst_feature}, R² = {best_r2:.3f}")
    else:
        break

print("\nFinal selected features:", selected)
print("Final R² Score:", best_r2)

# Train final model
final_model = LinearRegression().fit(X[selected], y)
df["Predicted_Price"] = final_model.predict(X[selected])
print("\nPredicted Price:\n", df)
```

```python
# TASK 19
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
data = {
    "Study_Hours": [2, 4, 6, 8, 10, 12],
    "Sleep_Hours": [7, 6, 5, 8, 7, 6],
    "Attendance": [60, 65, 70, 75, 80, 85],
    "Assignments_Completed": [3, 4, 5, 6, 7, 8],
    "Completion_Time": [90, 85, 80, 75, 70, 65]
}
df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Completion_Time", axis=1)
y = df["Completion_Time"]

# Backward Elimination
selected = list(X.columns)  # start with all features
best_r2 = r2_score(y, LinearRegression().fit(X[selected], y).predict(X[selected]))

while len(selected) > 1:  # stop if only one feature remains
    scores = {}
    for feature in selected:
        trial = selected.copy()
        trial.remove(feature)  # test model without this feature
        if len(trial) == 0:   # skip if no feature left
            continue
        model = LinearRegression().fit(X[trial], y)
        r2 = r2_score(y, model.predict(X[trial]))
        scores[feature] = r2

    if not scores:  # break if no scores calculated
        break

    worst_feature = min(scores, key=scores.get)
    worst_score = scores[worst_feature]

    if worst_score >= best_r2:  # remove if R² improves or stays same
        selected.remove(worst_feature)
        best_r2 = worst_score
        print(f"Removed: {worst_feature}, R² = {best_r2:.3f}")
    else:
        break

print("\nFinal selected features:", selected)
print("Final R² Score:", best_r2)

# Train final model with selected features
final_model = LinearRegression().fit(X[selected], y)

# Predict Completion Time
df["Predicted_Completion_Time"] = final_model.predict(X[selected])
print("\nPredicted Completion Time:\n", df)
```

```python
#Recursive Feature Elimination
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import RFE
data = {
    "Size_sqft": [1000, 1200, 1500, 1800, 2000, 2200],
    "Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [10, 5, 8, 12, 15, 20],
    "Distance_to_City": [5, 7, 4, 6, 3, 8],
    "Price": [200000, 250000, 280000, 300000, 350000, 370000]
}
df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Price", axis=1)
y = df["Price"]

# Apply Recursive Feature Elimination (RFE)
model = LinearRegression()
rfe = RFE(model, n_features_to_select=2)  # keep only 2 best features
rfe.fit(X, y)                             #dropping weaker features

# Get selected features
selected_features = X.columns[rfe.support_]   #a boolean
print("Selected Features:", list(selected_features))

# Train final model with selected features
final_model = LinearRegression().fit(X[selected_features], y)

# Predict house prices
df["Predicted_Price"] = final_model.predict(X[selected_features])
print("\nPredicted House Prices:\n", df)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import RFE
data = {
    "Engine_Size": [1000, 1200, 1500, 1600, 1800, 2000, 2200, 2500],
    "Horsepower": [70, 80, 100, 110, 130, 150, 170, 200],
    "Car_Weight": [800, 950, 1100, 1200, 1300, 1400, 1500, 1600],
    "Top_Speed": [140, 150, 160, 170, 180, 190, 200, 220],
    "Fuel_Efficiency": [22, 21, 19, 18, 17, 15, 14, 12]  # Target variable
}
df = pd.DataFrame(data)

# Features and target
X = df.drop("Fuel_Efficiency", axis=1)
y = df["Fuel_Efficiency"]

# Apply RFE
model = LinearRegression()
rfe = RFE(model, n_features_to_select=2)
fit = rfe.fit(X, y)

# Selected features
selected_features = X.columns[fit.support_].tolist()
print("Selected Features for Fuel Efficiency Prediction:", selected_features)

# Build model with selected features only
X_selected = X[selected_features]
model.fit(X_selected, y)

# Predictions
df["Predicted_Fuel_Efficiency"] = model.predict(X_selected)
print("\nActual vs Predicted Fuel Efficiency:")
print(df[["Fuel_Efficiency", "Predicted_Fuel_Efficiency"]])
```
