```python
#Creating Vectors
import numpy as np

# Define vectors
v1 = np.array([2, 3, 5])
v2 = np.array([1, 0, 4])

print("Vector v1:", v1)
print("Vector v2:", v2)
```

```python
#Vector Addition & Subtraction
# Addition
add = v1 + v2
print("v1 + v2 =", add)

# Subtraction
sub = v1 - v2
print("v1 - v2 =", sub)
```

```python
#Scalar Multiplication
scalar_mult = 3 * v1
print("3 * v1 =", scalar_mult)
```

```python
#Dot Product
#v1*v2=∑(v1i×v2i)
dot = np.dot(v1, v2) # 2*1 + 3*0 +5*4 =22
print("Dot Product:", dot)
```

```python
#Cross Product (3D only)


cross = np.cross(v1, v2) #   3*4 -5*0=12  5*1 - 4*2  = -3  2*0 - 3 =-3
print("Cross Product:", cross)
```

```python
#Magnitude (Norm)
magnitude_v1 = np.linalg.norm(v1)
print("Magnitude of v1:", magnitude_v1)
```

```python
#Unit Vector (Normalization)
unit_v1 = v1 / np.linalg.norm(v1)
print("Unit Vector of v1:", unit_v1)
```

```python
#Angle Between Two Vectors
cos_theta = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
theta = np.degrees(np.arccos(cos_theta))
print("Angle between v1 and v2:", theta, "degrees")
```

```python
#Check Orthogonality
#Two vectors are orthogonal if their dot product is 0.

is_orthogonal = np.dot(v1, v2) == 0
print("Are v1 and v2 orthogonal?", is_orthogonal)
```

```python
#Maximum and Minimum Elements
x = np.array([3, 8, 1, 6, 2])
print("Maximum element:", np.max(x))
print("Minimum element:", np.min(x))
```

```python
#Distance Between Two Vectors
p = np.array([2, -1, 3]) 
q = np.array([5, 2, -1])
distance = np.linalg.norm(p - q)
print("Euclidean distance:", distance)
```

```python
#Angle Classification
def classify_angle(u, v):
    cos_theta = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
    theta = np.degrees(np.arccos(np.clip(cos_theta, -1.0, 1.0)))  # avoid rounding errors
    
    if np.isclose(theta, 90):
        return f"Right angle ({theta:.2f}°)"
    elif theta < 90:
        return f"Acute angle ({theta:.2f}°)"
    else:
        return f"Obtuse angle ({theta:.2f}°)"

print(classify_angle(np.array([1,2]), np.array([2,3])))
print(classify_angle(np.array([1,0]), np.array([0,1])))
print(classify_angle(np.array([-1,0]), np.array([1,1])))
```

```python
# 2. Derivative : f(x) = x^3-6x^2+9x

import sympy as sp

x = sp.symbols('x')
f = x**3 - 6*x**2 + 9*x

# Derivative
derivative = sp.diff(f, x)
print("f(x) =", f)
print("f'(x) =", derivative)

#critical pont f′(x)=0
critical_points = sp.solve(derivative, x)
print("Critical points:", critical_points)

#Second derivative test : f′(x) > 0 :local minimum   f′(x) < 0 :local maximum
secondderivative = sp.diff(derivative, x)
print("f''(x) =", secondderivative)

for cp in critical_points:
    second = secondderivative.subs(x, cp)
    if second > 0:
        print(f"x={cp} → Minimum")
    elif second < 0:
        print(f"x={cp} → Maximum")
    else:
        print(f"x={cp} → Inconclusive")
```

# Numerical partial derivatives
def partial_x(f, x, y, h=1e-5): #h is very more number
    return (f(x+h, y) - f(x-h, y)) / (2*h)

def partial_y(f, x, y, h=1e-5):
    return (f(x, y+h) - f(x, y-h)) / (2*h)

print("∂f/∂x at (1,2) ≈", partial_x(f, 1, 2)) #point(1,2)
print("∂f/∂y at (1,2) ≈", partial_y(f, 1, 2))

#Gradient (∇f)
def gradient(f, x, y):
    return np.array([partial_x(f, x, y), partial_y(f, x, y)])

print("Gradient at (1,2):", gradient(f, 1, 2))

```python
import numpy as np 
def f(x, y): 
    return x**2 *y + 3*y**2
    
def partial_x(f, x, y, h=1e-5): 
    return (f(x+h, y) - f(x-h, y)) / (2*h)
def partial_y(f, x, y, h=1e-5):
    return (f(x, y+h) - f(x, y-h)) / (2*h)

print("∂f/∂x at (1,2) ≈", partial_x(f, 1, 2)) #point(1,2)
print("∂f/∂y at (1,2) ≈", partial_y(f, 1, 2))

#Gradient (∇f)
def gradient(f, x, y):
    return np.array([partial_x(f, x, y), partial_y(f, x, y)])

print("Gradient at (1,2):", gradient(f, 1, 2))
```
