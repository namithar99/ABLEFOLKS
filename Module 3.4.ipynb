```python
#DATA MANIPULATION WITH PANDAS
 #Handling missing values and duplicates

import pandas as pd
import numpy as np

# Create a sample DataFrame with missing values and duplicates
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan', 'Alice'],
    'Age': [25, 30, np.nan, 28, 40, 25],
    'Department': ['IT', 'Finance', 'HR', 'IT', 'Operations', 'IT'],
    'Salary': [55000, 65000, 70000, 60000, np.nan, 55000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
#Detect Missing Values
# Check for missing values
print(df.isnull().sum())

#Drop Missing Values
df_dropped_rows = df.dropna()
print(df_dropped_rows)

df_dropped_rows = df.dropna(axis=1)
df_dropped_rows  #missing value with columns are removed

#Fill missing values with the column mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)
```

```python
#Handling Duplicates
#Detect Duplicates
# Check for duplicate 
duplicates = df.duplicated()
print(duplicates)

#drop duplicates
# Drop duplicate rows, keeping the first occurrence
df_no_duplicates = df.drop_duplicates()
print(df_no_duplicates)
```

```python
import pandas as pd
#merging
# Creating datasets
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Score': [85, 90, 95]
})

# Merge on a common column
merged = pd.merge(df1, df2)  # Inner Join
print(merged)

#joining
df3 = pd.DataFrame({
    'Score': [85, 90, 95],
    'Grade': ['B', 'A', 'A']})

df4 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie']})

# Join on index
joined = df4.join(df3)
print(joined)

#concatenation
df5 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Score': [85, 90]})
df6 = pd.DataFrame({'Name': ['Charlie', 'David'], 'Score': [95, 80]})

# Concatenate row-wise
concat_rows = pd.concat([df5, df6])
print(concat_rows)

concat_columns = pd.concat([df5, df6],axis=1)
print(concat_columns)
```

```python

import pandas as pd

url = "C:/Users/hp/Desktop/ABLEFOLKS/titanic.csv"
titanic_data = pd.read_csv(url)

print(titanic_data.head())
print("\nMissing values:\n", titanic_data.isnull().sum())
```

```python
# Number of rows in the DataFrame
print("Number of rows:", len(titanic_data))

# Shape of the DataFrame (rows, columns)
print("Dimensions:", titanic_data.shape)

# Count of non-missing values per column
print("\nNon-missing counts per column:")
print(titanic_data.count())
```

```python
import pandas as pd


# Handle missing values
# Fill Age with median
titanic_data['Age'].fillna(titanic_data['Age'].median(), inplace=True)

# Fill Embarked with mode (most frequent value)
titanic_data['Embarked'].fillna(titanic_data['Embarked'].mode()[0], inplace=True)

# Convert Cabin into Deck (first letter) and fill missing with 'U'
titanic_data['Deck'] = titanic_data['Cabin'].astype(str).str[0]
titanic_data['Deck'].replace("n", "U", inplace=True)  # 'n' comes from "nan"

# (Optional) Drop Cabin if you don’t want the detailed cabin numbers
titanic_data.drop(columns=['Cabin'], inplace=True)

# Show missing values after cleaning
print("\nMissing values after cleaning:\n", titanic_data.isnull().sum())

# Example analysis: average fare by class
avg_fare_by_class = titanic_data.groupby('Pclass')['Fare'].mean()
print("\nAverage Fare by Class:\n", avg_fare_by_class)
```

```python
#top 5 passengers by Fare (highest).
print(titanic_data.sort_values('Fare', ascending=False).head(5))
```

```python
#Compute the average Fare by Pclass.
print(titanic_data.groupby('Pclass')['Fare'].mean())

#Find the maximum Age by Embarked port.
print(titanic_data.groupby('Embarked')['Age'].max())

#Find the survival rate by Gender.
print(titanic_data.groupby('Sex')['Survived'].mean())

#Find the survival rate by Pclass.
print(titanic_data.groupby('Pclass')['Survived'].mean())
```

```python
#Compare survival of children (Age < 16) vs adults.

titanic_data['Child'] = titanic_data['Age'] < 16
print(titanic_data.groupby('Child')['Survived'].mean())
```

```python
class NegativeNumberError(Exception):
    pass

def square_root(x):
    if x < 0:
        raise NegativeNumberError("Cannot take square root of negative number!")
    return x ** 0.5

try:
    print(square_root(-9))
except NegativeNumberError as e:
    print("❌ Error:", e)
```
