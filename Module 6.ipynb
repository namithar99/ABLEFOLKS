```python
# Traditional Programming
def fruit_classifier(weight):
    if weight > 145:
        return "Apple"
    else:
        return "Orange"
print(fruit_classifier(170))   
print(fruit_classifier(130))   
```

```python
# Machine Learning Programming 
training_data = [
    {"features": [150], "label": "Apple"},
    {"features": [170], "label": "Banana"},
    {"features": [140], "label": "Orange"},
    {"features": [130], "label": "Grape"}
]

model = training_data  
def predict(weight):
    for item in model:
        if item["features"][0] == weight:
            return item["label"]
    return "Unknown"
print(predict(155))   
print(predict(130))   
```

```python
def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False
print(is_even(10))  
print(is_even(7))   
```

```python
training_data = [
    {"features": [1], "label": "Small"},
    {"features": [10], "label": "Medium"},
    {"features": [100], "label": "Large"}
]
model = training_data
def predict_size(number):
    for item in model:
        if item["features"][0] == number: # 0: only one feature
            return item["label"]
    return "Unknown"

print(predict_size(10))  
print(predict_size(50))  
```

```python
#Supervised 
training_data = [
    {"features": [85], "label": "Pass"},
    {"features": [70], "label": "Pass"},
    {"features": [40], "label": "Fail"},
    {"features": [30], "label": "Fail"}
]
def predict(score):
    for item in training_data:
        if item["features"][0] == score:  
            return item["label"]
    return "Unknown"   
print(predict(70))  
print(predict(40))   
print(predict(55)) 
```

```python
import pandas as pd
df = pd.DataFrame({
    "Student": ["A","B","C","D"],
    "Marks": [85, 70, 40, 30],
    "Result": ["Pass", "Pass", "Fail", "Fail"]
})
def predict_result(score):
    if score >= 50:
        return "Pass"
    else:
        return "Fail"
df["Predicted_Result"] = df["Marks"].apply(predict_result)
print("Dataset with Predictions:\n", df)
new_scores = [55, 45]
for score in new_scores:
    print(f"New student with marks {score} → Result: {predict_result(score)}")
```

```python
import pandas as pd
df = pd.DataFrame({
    "Student": ["A","B","C","D","E","F","G","H"],
    "Marks": [35, 45, 55, 60, 65, 75, 85, 95]
})

# Grouping function (clustering rule)
def group_marks(marks):
    if marks < 50:
        return "Low"
    elif 50 <= marks < 70:
        return "Medium"
    else:
        return "High"
# Assign clusters to dataset
df["Cluster"] = df["Marks"].apply(group_marks)
print("Clustered Data:\n", df)

new_score = 68
predicted_cluster = group_marks(new_score)
print(f"\nNew Student with score {new_score} → Cluster: {predicted_cluster}")
```

```python
df = pd.DataFrame({
    "Person": ["P1","P2","P3","P4","P5"],
    "BMI": [30, 25, 22, 27, 20],
    "Steps": [2000, 5000, 8000, 6000, 12000]
})

# Grouping function
def group_health(row):
    if row["BMI"] > 28 and row["Steps"] < 4000:
        return "Unfit"
    elif row["Steps"] < 7000:
        return "Moderate"
    else:
        return "Fit"
df["Cluster"] = df.apply(group_health, axis=1)
print(df)

new_person = {"BMI": 26, "Steps": 8500}
predicted_cluster = group_health(new_person)
print(f"\nNew person → Cluster: {predicted_cluster}")
```

```python
#TASK 4
df = pd.DataFrame({
    "Student": ["A","B","C","D","E"],
    "Math": [35, 60, 75, 90, 55],
    "English": [40, 65, 70, 85, 50]
})

def group_student(row):
    avg = (row["Math"] + row["English"]) / 2
    if avg < 50:
        return "Weak"
    elif avg < 75:
        return "Average"
    else:
        return "Strong"

df["Cluster"] = df.apply(group_student, axis=1)
print(df)
new_student = {"Math": 72, "English": 80}
print(f"New student → Cluster: {group_student(new_student)}")
```

```python
# Reinforcement
position = 0
goal = 5
reward = 0
actions = ["right", "right", "left", "right", "right"]

for action in actions:
    if action == "right":
        position += 1
        reward += 1
    elif action == "left":
        position -= 1
        reward -= 1
    print(f"Action: {action}, Position: {position}, Reward: {reward}")
print("Final Position:", position)
print("Total Reward:", reward)
```

```python
position = 0
coins = [2, 5]       # reward positions
obstacles = [3]      # penalty positions
reward = 0

# Predefined sequence of actions
actions = ["right", "right", "right", "left", "right", "right"]

print("Starting at Position:", position)

# Loop through each action
for action in actions:
    if action == "right":
        position += 1
    elif action == "left":
        position -= 1
    
    # Rewards & Penalties
    if position in coins:
        reward += 5
        coins.remove(position)   # coin collected (disappears)
        print(f"Collected a coin at {position}! Reward +5")
    elif position in obstacles:
        reward -= 5
        print(f"Hit an obstacle at {position}! Penalty -5")
    else:
        reward -= 1
        print(f"Moved to {position}, no coin/obstacle. Penalty -1")
    
    print(f"Current Position: {position}, Total Reward: {reward}\n")
print("Final Reward:", reward)
```

```python
# TASK 5
position = 0
coins = [2, 4, 6, 8]   # 4 coins on the line
reward = 0

actions = ["right", "right", "right", "right", "right", "right", "right", "right"]

print("Starting at Position:", position)

# Loop through each action
for action in actions:
    if action == "right":
        position += 1
    else:
        position -= 1

    # Check rewards
    if position in coins:
        reward += 2
        coins.remove(position)  # coin disappears
        print(f"Collected a coin at {position}! Reward +2")
    else:
        reward -= 1
        print(f"Moved to {position}, no coin found. Penalty -1")
    
    print(f"Current Position: {position}, Total Reward: {reward}\n")
print("Final Reward:", reward)
```

```python
# TASK 6
position = 0
coins = [2, 5, 7]        # 3 coin positions
obstacles = [3, 6]       # 2 obstacle positions
reward = 0

actions = ["right", "right", "right", "right", "right", "right", "right"]
print("Starting at Position:", position)

# Loop through each action
for action in actions:
    if action == "right":
        position += 1
    elif action == "left":
        position -= 1

    # Rewards and penalties
    if position in coins:
        reward += 5
        coins.remove(position)   # coin disappears after collection
        print(f"Collected a coin at {position}! Reward +5")
    elif position in obstacles:
        reward -= 5
        print(f"Hit an obstacle at {position}! Penalty -5")
    else:
        reward -= 1
        print(f"Moved to {position}, nothing found. Penalty -1")
    
    print(f"Current Position: {position}, Total Reward: {reward}\n")
print("Final Reward:", reward)
```
