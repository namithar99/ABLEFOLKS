```python
#Reusable & Modular Code
# circumference :name of the function & radius is input paramter

def circumference(radius):
    return 2 * 3.14 * radius

print(circumference(2))
print(circumference(10))
```

```python
#Temperature Converter

def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9

print(celsius_to_fahrenheit(25))  # 77.0
print(fahrenheit_to_celsius(98.6))  # 37.0
```

```python
#Prime

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):               # (4^1/2+ 1 =  3)
        if num % i == 0:                                # % gives the remainder after dividing num by i.
            return False                                # i divide num exactly without a remainder (not a prime)
    return True

print(is_prime(7))   
print(is_prime(20))  
```

```python
#factorial
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i                    # result = result * i 2!  i=1 1+1=2 2*1
    return result

print(factorial(5)) 
```

```python
#Example of Reusable + Modular Code
# Function to calculate average

def average(marks):
    return sum(marks) / len(marks)

# Function to assign grade
def grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "F"

# Main program
marks = [85, 78, 92, 20]

avg_marks = average(marks)
print("Average:", avg_marks)
print("Grade:", grade(avg_marks))
```

```python
#BMI Calculator (BMI = Weight (kg) ÷ Height² (m²).)
def calculate_bmi(weight_kg, height_m):
    """Return BMI value."""
    return weight_kg / (height_m ** 2)

def bmi_category(bmi):
    """Return BMI category."""
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

# Example usage
bmi = calculate_bmi(70, 1.75)
print("BMI:", bmi)
print("Category:", bmi_category(bmi))
```

```python
## Mini ATM Simulation (Using Functions)

# Function to display menu
def display_menu():
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")

# Function to check balance
def check_balance(balance):
    print(f"Your current balance is ₹{balance}")

# Function to deposit money
def deposit(balance, amount):
    if amount <= 0:
        print("Invalid deposit amount!")
    else:
        balance += amount
        print(f"₹{amount} deposited successfully.")
    return balance

# Function to withdraw money
def withdraw(balance, amount):
    if amount <= 0:
        print("Invalid withdrawal amount!")
    elif amount > balance:
        print("Insufficient balance!")
    else:
        balance -= amount
        print(f"₹{amount} withdrawn successfully.")
    return balance

# Main Program
def atm():
    balance = 5000  # Initial balance
    while True:
        display_menu()
        choice = input("Enter your choice: ")

        if choice == "1":
            check_balance(balance)
        elif choice == "2":
            amt = float(input("Enter amount to deposit: ₹"))
            balance = deposit(balance, amt)
        elif choice == "3":
            amt = float(input("Enter amount to withdraw: ₹"))
            balance = withdraw(balance, amt)
        elif choice == "4":
            print("Thank you for using our ATM!")
            break
        else:
            print("Invalid choice, please try again.")

# Run the ATM program
atm()
```
