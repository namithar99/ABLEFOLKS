```python
# TASK

import csv

file_path = "C:/Users/hp/Downloads/movies_info.csv"  # Change this to your dataset path

# Variables for all tasks
total_movies = 0
genres_set = set()
genre_count = {}
longest_movie = ""
max_length = 0
keyword = "love"
matching_movies = []

# Open the file once
with open(file_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)  # Read using column names
    for row in reader:
        total_movies += 1  # Q1: Count total movies

# Q2 & Q3: Process genres
        genres = row["genres"].split(",")
        for g in genres:
            g = g.strip()
            genres_set.add(g)  # Unique genres
            genre_count[g] = genre_count.get(g, 0) + 1  # Count per genre

# Q4: Find longest description
        description = row["overview"]
        if len(description) > max_length:
            max_length = len(description)
            longest_movie = row["original_title"]

# Q5: Search for keyword
        if keyword.lower() in description.lower():
            matching_movies.append(row["original_title"])

# Printing results
print("1. Total Movies:", total_movies)
print("2. Unique Genres:", genres_set)
print("3. Movies per Genre:", genre_count)
print("4. Longest Description Movie:", longest_movie)
print("   Description Length:", max_length)
print(f'5. Movies containing "{keyword}":', matching_movies)
```

```python
import csv

unique_genres = set() #set automatically removes duplicates.

with open("C:/Users/hp/Downloads/movies_info.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)

    header = next(reader)              # Read header row
    genre_index = header.index("genres")  # Find the column number of 'genres',2nd column, then genre_index = 1.

    for row in reader:
        genres = row[genre_index]                            # extract genres column
        if genres and genres != "[]":                        #if the value is not empty and not just "[]" (meaning no genres).
            genres = genres.strip("[]").replace("'", "")     # .strip("[]") → removes the square brackets [ ]#replace("'", "") → removes single quotes '.
            for g in genres.split(","):                      #.split(",") → splits the string into individual genres
                unique_genres.add(g.strip())                 #.g.strip() → removes any extra spaces

# Show result
print("Unique genres in the dataset:")
#print(sorted(unique_genres))
print(unique_genres)
```

```python
import csv

genre_counts = {}   # empty dictionary

with open("C:/Users/hp/Downloads/movies_info.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)
    genre_index = header.index("genres")

    for row in reader:
        genres = row[genre_index]
        if genres and genres != "[]":
            genres = genres.strip("[]").replace("'", "")
            for g in genres.split(","):
                genre = g.strip()
                if genre:   
                    if genre in genre_counts:
                        genre_counts[genre] += 1
                    else:
                        genre_counts[genre] = 1

# print results
print("Movies per genre:")
for genre, count in genre_counts.items():
    print(f"{genre}: {count}")
```

```python
#3.country with the largest area

largest_area = countries[0]

for c in countries:
    if c["area"] > largest_area["area"]:
        largest_area = c

print(largest_area["name"]["common"], "-", largest_area["area"])
```

```python
#4.Create a dictionary with region as the key and list of country names as values

region_dict = {}

for c in countries:
    region = c["region"]            # get region
    name = c["name"]["common"]      # get country name
    if region not in region_dict:   #region is already in the dictionary.#new entry with an empty list.:if not
        region_dict[region] = []
    
    # add the country name to the list
    region_dict[region].append(name)

# print result
for region in region_dict:
    print(region, ":", region_dict[region])
```

```python
#Create a dictionary with region as the key and list of country names as values

region_dict = {}
for c in countries:
    region = c.get('region', 'Unknown')
    region_dict.setdefault(region, []).append(c['name']['common'])
print(region_dict)
```
